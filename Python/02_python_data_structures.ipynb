## 1. Lists

A list is an ordered and mutable collection of elements in Python.

Lists can store multiple values and can contain elements of different data types.

Lists are widely used in data processing and are an important foundation for working with datasets.

````python
marks = [78, 85, 92, 74, 81]

print("Marks:", marks)
print("Number of students:", len(marks))
````

### 1.1 List Indexing

Each element in a list has an index.

Python uses zero-based indexing, meaning the first element has index `0`.

Negative indexing can be used to access elements from the end of the list.

````python
marks = [78, 85, 92, 74, 81]

print("First mark:", marks[0])
print("Third mark:", marks[2])
print("Last mark:", marks[-1])
````

### 1.2 List Slicing

List slicing is used to extract a specific portion of a list.

Basic syntax:

`list[start:stop]`

The `start` index is included, while the `stop` index is excluded.

````python
marks = [78, 85, 92, 74, 81]

print("First three marks:", marks[0:3])
print("Middle marks:", marks[1:4])
print("Last two marks:", marks[-2:])
````

#### Slicing with Step

A third value can be used in slicing to specify the step size.

The syntax is:

`list[start:stop:step]`

When start and stop are omitted, the slicing can begin from the start of the list and continue to the end.

````python
marks = [78, 85, 92, 74, 81]

print("Every second mark:", marks[::2])
````

### 1.3 Modifying Lists

Lists are mutable, which means their elements can be changed after the list is created.

Common operations include:

- Updating an element
- Adding an element using `append()`
- Inserting an element using `insert()`
- Removing an element using `remove()`
- Removing an element by index using `pop()`

````python
marks = [78, 85, 92, 74, 81]

marks[3] = 76

print("Updated marks:", marks) 
````

#### Adding an Element with append()

The `append()` method adds a new element to the end of a list.

````python
marks = [78, 85, 92, 76, 81]

marks.append(88)

print("Marks after adding a new value:", marks)
````

#### Adding an Element with insert()

The `insert()` method adds an element at a specified index.

Syntax:

`list.insert(index, value)`

````python
marks = [78, 85, 92, 76, 81]

marks.insert(2, 88)

print("Marks after insertion:", marks)
````

#### Removing an Element with remove()

The `remove()` method removes the first occurrence of a specified value from a list.

````python
marks = [78, 85, 88, 92, 76, 81]

marks.remove(88)

print("Marks after removing 88:", marks)
````

#### Removing an Element with pop()

The `pop()` method removes and returns an element from a list using its index.

If no index is provided, `pop()` removes the last element.

````python
marks = [78, 85, 92, 76, 81]

removed_mark = marks.pop(2)

print("Removed mark:", removed_mark)
print("Marks after pop:", marks)
````

#### Traversing a List with a for Loop

A `for` loop can be used to visit and process each element of a list one by one.

````python
marks = [78, 85, 92, 76, 81]

for mark in marks:
    print("Mark:", mark)
````

#### Filtering Values from a List

A list can be processed using conditions to select values that satisfy a specific requirement.

````python
marks = [45, 78, 92, 61, 35, 88, 49]

passing_marks = []

for mark in marks:
    if mark >= 50:
        passing_marks.append(mark)

print("Original marks:", marks)
print("Passing marks:", passing_marks)
````

#### Transforming Values in a List

List elements can be transformed by applying an operation to each value.

This is a basic example of data transformation, where the original values are converted into a new form.

````python
marks = [45, 78, 92, 61, 35, 88, 49]

bonus_marks = []

for mark in marks:
    bonus_marks.append(mark + 5)

print("Original marks:", marks)
print("Marks after bonus:", bonus_marks)
````

#### Useful List Operations

Python provides built-in functions and list methods for common data-processing tasks.

Some useful operations include:

- `len()` → number of elements
- `sum()` → total of numeric values
- `min()` → smallest value
- `max()` → largest value
- `sort()` → sort the list

````python
marks = [78, 85, 92, 76, 81]

print("Number of marks:", len(marks))
print("Total marks:", sum(marks))
print("Minimum mark:", min(marks))
print("Maximum mark:", max(marks))

marks.sort()
print("Sorted marks:", marks)
````

### 1.12 Practical Example: Student Marks Analysis

Lists can be used to store and analyze data such as student marks.

In this example, we will calculate basic statistics from a list of marks and identify students who meet a given performance requirement.

````python
marks = [45, 78, 92, 61, 35, 88, 49, 76, 91, 67]

total_marks = sum(marks)
average_marks = total_marks / len(marks)
highest_mark = max(marks)
lowest_mark = min(marks)

passing_marks = []

for mark in marks:
    if mark >= 50:
        passing_marks.append(mark)

print("Total marks:", total_marks)
print("Average marks:", average_marks)
print("Highest mark:", highest_mark)
print("Lowest mark:", lowest_mark)
print("Number of passing students:", len(passing_marks))
print("Passing marks:", passing_marks)
````

### Challenge 1 — Student Performance Analysis

You are given the marks of 10 students:

marks = [45, 78, 92, 61, 35, 88, 49, 76, 91, 67]

### Tasks

1. Find the average marks.
2. Count how many students scored above the average.
3. Create a new list containing only the marks above the average.
4. Find the highest mark among those students.
5. Find how many students scored between 50 and 80 inclusive.
6. Determine whether more students scored above or below the average.

### Rules

- Use Python lists and loops.
- You may use len(), sum(), min(), and max().
- Do not use NumPy or Pandas.
- Do not hard-code the answers.
- The solution should work if the marks list changes.

````python
marks = [45, 78, 92, 61, 35, 88, 49, 76, 91, 67]

# 1. Calculate average marks
average_marks = sum(marks) / len(marks)

# 2 & 3. Find students above average
above_average_count = 0
below_or_equal_average_count = 0
above_average_marks = []

for mark in marks:
    if mark > average_marks:
        above_average_count += 1
        above_average_marks.append(mark)
    else:
        below_or_equal_average_count += 1

# 4. Find highest mark among students above average
highest_above_average = max(above_average_marks)

# 5. Count students scoring between 50 and 80 inclusive
between_50_and_80 = 0

for mark in marks:
    if mark >= 50 and mark <= 80:
        between_50_and_80 += 1

# 6. Compare above-average and below/equal-average students
if above_average_count > below_or_equal_average_count:
    comparison = "More students scored above the average."
elif below_or_equal_average_count > above_average_count:
    comparison = "More students scored below or equal to the average."
else:
    comparison = "The number of students is equal on both sides of the average."

# Display results
print("Average marks:", average_marks)
print("Students above average:", above_average_count)
print("Marks above average:", above_average_marks)
print("Highest mark among above-average students:", highest_above_average)
print("Students scoring between 50 and 80:", between_50_and_80)
print("Comparison:", comparison)
````

### 1.14 List Comprehension

List comprehension provides a concise way to create a new list by applying an expression to each element of an existing iterable.

It can often replace a simple `for` loop used for creating a new list.

Basic structure:

`[expression for item in iterable]`

List comprehensions are commonly used in Python for data transformation and filtering.

````python
marks = [45, 78, 92, 61, 35, 88, 49]

bonus_marks = [mark + 5 for mark in marks]

print("Marks after bonus:", bonus_marks)
````

### 1.15 Conditional List Comprehension

A condition can be added to a list comprehension to include only the elements that satisfy a specific requirement.

Basic structure:

`[expression for item in iterable if condition]`

This is useful for concise filtering and transformation of data.

````python
marks = [45, 78, 92, 61, 35, 88, 49]

passing_marks = [mark for mark in marks if mark >= 50]

print("Passing marks:", passing_marks)
````

### 1.16 Filtering and Transforming with List Comprehension

A list comprehension can both filter values using a condition and transform the selected values using an expression.

This pattern is useful for creating a processed dataset from existing data.

````python
marks = [45, 78, 92, 61, 35, 88, 49]

bonus_passing_marks = [mark + 5 for mark in marks if mark >= 50]

print("Passing marks after bonus:", bonus_passing_marks)
````

````python
marks = [40, 50, 65, 79, 85]

result = [mark + 10 for mark in marks if mark >= 60]

print("Result:", result)
````

## 2. Tuples

### What is a Tuple?

A tuple is an ordered and immutable collection of elements in Python.

Tuples can store multiple values and can contain elements of different data types.

Unlike lists, the elements of a tuple cannot be modified after the tuple is created.

````python
student = ("Divyanshu", 22, "MCA")

print("Student:", student)
print("Number of elements:", len(student))
````

### Accessing Tuple Elements

Tuple elements are accessed using their index.

- Positive indexing starts from `0`.
- Negative indexing starts from `-1`.
- Tuples support indexing in the same way as lists.

````python
student = ("Divyanshu", 22, "MCA", "Data Science")

print("First element:", student[0])
print("Second element:", student[1])
print("Last element:", student[-1])
print("Second last element:", student[-2])
````

### Tuple Slicing

Tuple slicing is used to extract a portion of a tuple.

The basic syntax is:

`tuple[start:stop:step]`

The `stop` index is excluded from the result.

````python
student = ("Divyanshu", 22, "MCA", "Data Science", "India")

print("First three elements:", student[:3])
print("From second element:", student[1:])
print("Every second element:", student[::2])
print("Reversed tuple:", student[::-1])
````

### Tuple Immutability

Tuples are immutable, which means their existing elements cannot be changed, added, or removed after the tuple is created.

This is one of the main differences between a list and a tuple.

````python
student = ("Divyanshu", 22, "MCA")

print("Original tuple:", student)

# Trying to modify an existing element
# student[1] = 23  #THIS LINE GENERATE THE ERROR, because tuple ke existing element ko modify nahi kar sakte.
````

````python
student = ("Divyanshu", 22, "MCA")

print("Original tuple:", student)

# A new tuple can be created instead of modifying the existing one
updated_student = ("Divyanshu", 23, "MCA")

print("New tuple:", updated_student)
````

### Tuple Operations

Tuples support several common operations such as:

- Concatenation using `+`
- Repetition using `*`
- Membership checking using `in`
- Finding the number of elements using `len()`
- Counting an element using `count()`
- Finding the position of an element using `index()`

````python
numbers = (10, 20, 30, 20, 40)

print("Length:", len(numbers))
print("Count of 20:", numbers.count(20))
print("Index of 30:", numbers.index(30))

print("Is 40 present?", 40 in numbers)
print("Is 50 present?", 50 in numbers)

first = (1, 2, 3)
second = (4, 5)

print("Concatenated tuple:", first + second)
print("Repeated tuple:", first * 2)
````

## 3. Sets

A set is an unordered and mutable collection of unique elements in Python.

Sets automatically remove duplicate values and are useful when we need to work with unique data.


````python
numbers = {10, 20, 30, 20, 40, 10}

print("Set:", numbers)
print("Number of unique elements:", len(numbers))
````

### 3.1 Adding Elements to a Set

The `add()` method is used to add a single element to a set.

Since sets contain only unique elements, adding an element that already exists does not create a duplicate.

````python
numbers = {10, 20, 30}

numbers.add(40)
numbers.add(20)

print("Set after adding elements:", numbers)
````

### 3.2 Removing Elements from a Set

The `remove()` method removes a specified element from a set.

The `discard()` method also removes an element, but it does not raise an error if the element is not present.


````python
numbers = {10, 20, 30, 40}

numbers.remove(30) #if number are not available gives us an error
numbers.discard(50) # number are available or not it doesn't gives an error

print("Set after removal:", numbers)
````

### 3.3 Membership in a Set

The `in` operator is used to check whether an element exists in a set.

The `not in` operator checks whether an element does not exist in a set.

````python
numbers = {10, 20, 30, 40}

print("Is 20 present?", 20 in numbers)
print("Is 50 present?", 50 in numbers)
print("Is 50 absent?", 50 not in numbers)
````

### 3.4 Set Union

Union combines all unique elements from two sets.

It can be performed using the `|` operator or the `union()` method.

````python
python_students = {"Aman", "Riya", "Rahul"}
data_students = {"Riya", "Rahul", "Neha"}

all_students = python_students | data_students # | means union()
print("All students:", all_students)
````

### 3.5 Set Intersection

Intersection returns only the elements that are common to both sets.

It can be performed using the `&` operator or the `intersection()` method.

````python
python_students = {"Aman", "Riya", "Rahul"}
data_students = {"Riya", "Rahul", "Neha"}

common_students = python_students & data_students

print("Students in both courses:", common_students)
````

### 3.6 Set Difference

Set difference returns the elements that are present in the first set but not in the second set.

It can be performed using the `-` operator or the `difference()` method.

````python
python_students = {"Aman", "Riya", "Rahul", "Neha"}
data_students = {"Riya", "Rahul", "Priya"}

python_only = python_students - data_students
data_only = data_students - python_students

print("Only in Python course:", python_only)
print("Only in Data Science course:", data_only)
````

### 3.7 Set Operations — Quick Comparison

| Operation | Symbol | Meaning |
|---|---|---|
| Union | '\|' | All unique elements from both sets |
| Intersection | `&` | Common elements |
| Difference | `-` | Elements present only in the first set |

These operations are useful for comparing groups of unique values.

### 3.8 Practical Example: Finding Unique Values

Sets are useful when a dataset contains repeated values and we only need the unique values.

For example, suppose we have student course registrations with duplicate entries.

````python
courses = [
    "Python",
    "Data Science",
    "Python",
    "Machine Learning",
    "Data Science",
    "Python"
]

unique_courses = set(courses)

print("Original courses:", courses)
print("Unique courses:", unique_courses)
print("Number of unique courses:", len(unique_courses))
````

## 4. Dictionaries

A dictionary is a mutable collection of data stored in key-value pairs.

Each key is used to access its corresponding value.

Dictionaries are written using curly braces `{}`.

Example:

```python
student = {
    "name": "Divyanshu",
    "age": 22,
    "course": "MCA"
}
Here:
- "name", "age", and "course" are keys.
- "Divyanshu", 22, and "MCA" are their corresponding values.

````python

### 4.2 Creating and Accessing Dictionary Values

student = {
    "name": "Divyanshu",
    "age": 22,
    "course": "MCA",
    "specialization": "Data Science"
}

print("Student:", student)

print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])
````

### 4.3 Accessing Values Using `get()`

The `get()` method is used to access a value using its key.

Unlike direct indexing, `get()` returns `None` when the key does not exist instead of raising an error.

````python
student = {
    "name": "Divyanshu",
    "age": 22,
    "course": "MCA"
}

print("Name:", student.get("name"))
print("Course:", student.get("course"))
print("City:", student.get("city"))
````

### 4.4 Adding and Updating Dictionary Elements

New key-value pairs can be added to a dictionary by assigning a value to a new key.

If the key already exists, its value is updated.

````python
student = {
    "name": "Divyanshu",
    "age": 22,
    "course": "MCA"
}

# Adding a new key-value pair
student["city"] = "Bhopal"

# Updating an existing value
student["age"] = 23

print("Updated student:", student)
````

### 4.5 Removing Dictionary Elements

Dictionary elements can be removed using methods such as:

- `pop()` — removes the specified key and returns its value.
- `popitem()` — removes the last inserted key-value pair.
- `del` — removes a specified key-value pair.
- `clear()` — removes all elements from the dictionary.

````python
student = {
    "name": "Divyanshu",
    "age": 22,
    "course": "MCA",
    "city": "Bhopal"
}

removed_value = student.pop("city")

print("Removed value:", removed_value)
print("After pop():", student)

student["city"] = "Bhopal"
student.popitem()

print("After popitem():", student)

del student["age"]

print("After del:", student)

student.clear()

print("After clear():", student)
````

### 4.6 Keys, Values and Items

Dictionaries provide methods to access their keys, values, and key-value pairs.

- `keys()` returns all keys.
- `values()` returns all values.
- `items()` returns all key-value pairs.

````python
student = {
    "name": "Divyanshu",
    "age": 22,
    "course": "MCA"
}

print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())
````

### 4.7 Dictionary Traversal

Dictionary traversal means accessing dictionary elements using a loop.

We can iterate through:

- Keys using `keys()`
- Values using `values()`
- Key-value pairs using `items()`

The `items()` method is useful when both keys and values are required.

````python
student = {
    "name": "Divyanshu",
    "age": 22,
    "course": "MCA",
    "specialization": "Data Science"
}

# Traversing keys
for key in student.keys():
    print(key)

# Traversing values
for value in student.values():
    print(value)

# Traversing key-value pairs
for key, value in student.items():
    print(key, ":", value)
````

### 4.8 Practical Example: Student Marks Analysis

Dictionaries are useful for storing structured data.

In this example, student names are stored as keys and
their corresponding marks are stored as values.

We will calculate:

- Total marks
- Average marks
- Highest marks
- Student with the highest marks

````python
student_marks = {
    "Aman": 85,
    "Riya": 92,
    "Rahul": 78,
    "Neha": 90,
    "Priya": 88
}

total = sum(student_marks.values())

average = total / len(student_marks)

highest = max(student_marks.values())

top_student = max(student_marks, key=student_marks.get)

print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Marks:", highest)
print("Top Student:", top_student)
````

## 5. Nested Data Structures

Common examples include:

- List of lists
- List of dictionaries
- Dictionary containing lists

### 5.2 List of Lists

A list can contain other lists as its elements.

This is useful for representing tabular or matrix-like data.

````python
marks = [
    [85, 90, 78],
    [92, 88, 95],
    [76, 84, 80]
]

print("Marks:", marks)

print("First student's marks:", marks[0])
print("First student's second subject mark:", marks[0][1])
````

### 5.3 List of Dictionaries

### 6.2 Filtering Records

Filtering means selecting only those records that satisfy a given condition.

For example, we can find students who scored 85 or more marks.

````python
students = [
    {"name": "Aman", "marks": 85},
    {"name": "Riya", "marks": 92},
    {"name": "Rahul", "marks": 78},
    {"name": "Neha", "marks": 90},
    {"name": "Priya", "marks": 88}
]

high_scorers = []

for student in students:
    if student["marks"] >= 85:
        high_scorers.append(student)

print("Students scoring 85 or more:")

for student in high_scorers:
    print(student["name"], ":", student["marks"])
````

### 6.3 Sorting Records

Sorting arranges records according to a specific value.

For example, student records can be sorted according to their marks.

````python
students = [
    {"name": "Aman", "marks": 85},
    {"name": "Riya", "marks": 92},
    {"name": "Rahul", "marks": 78},
    {"name": "Neha", "marks": 90},
    {"name": "Priya", "marks": 88}
]

sorted_students = sorted(students, key=lambda student: student["marks"], reverse=True)

print("Students sorted by marks:")

for student in sorted_students:
    print(student["name"], ":", student["marks"])
````

### 6.4 Summary Statistics

Summary statistics help us understand a dataset quickly.

For student marks, we can calculate:

- Total marks
- Average marks
- Highest marks
- Lowest marks
- Number of students

````python
students = [
    {"name": "Aman", "marks": 85},
    {"name": "Riya", "marks": 92},
    {"name": "Rahul", "marks": 78},
    {"name": "Neha", "marks": 90},
    {"name": "Priya", "marks": 88}
]

marks = [student["marks"] for student in students]

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

print("Number of students:", len(students))
print("Total marks:", total)
print("Average marks:", average)
print("Highest marks:", highest)
print("Lowest marks:", lowest)
````

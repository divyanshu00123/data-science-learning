Variables are used to store data in Python.

Python automatically determines the data type based on the assigned value.

### Type Conversion

Type conversion is the process of converting a value from one data type to another.

Common conversions include:

- `int()` → converts a value to integer
- `float()` → converts a value to floating-point number
- `str()` → converts a value to string
- `bool()` → converts a value to Boolean

````python
marks = "85"
marks_as_integer = int(marks)

percentage = 85
percentage_as_float = float(percentage)

student_id = 101
student_id_as_string = str(student_id)

print("Original marks:", marks, type(marks))
print("Converted marks:", marks_as_integer, type(marks_as_integer))

print("Percentage:", percentage_as_float, type(percentage_as_float))
print("Student ID:", student_id_as_string, type(student_id_as_string))
````

### Key Takeaways

- Python variables can store different types of data.
- Common basic data types include `str`, `int`, `float`, and `bool`.
- The `type()` function is used to identify the data type of a value.
- Type conversion allows values to be converted between compatible data types.
- Correct data types are important when working with real-world datasets.

## 2. Operators

Operators are symbols used to perform operations on values and variables.

In Python, common categories of operators include:

- Arithmetic operators
- Comparison operators
- Logical operators
- Assignment operators

### Arithmetic Operators

Arithmetic operators are used to perform mathematical calculations.

````python
a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
````

### Comparison Operators

Comparison operators compare two values and return a Boolean result (`True` or `False`).

They are commonly used in decision-making and data filtering.

````python
cgpa = 8.26
required_cgpa = 6.32

print("CGPA greater than required:", cgpa > required_cgpa)
print("CGPA less than required:", cgpa < required_cgpa)
print("CGPA equal to required:", cgpa == required_cgpa)
print("CGPA not equal to required:", cgpa != required_cgpa)
print("CGPA greater than or equal to required:", cgpa >= required_cgpa)
print("CGPA less than or equal to required:", cgpa <= required_cgpa)
````

### Logical Operators

Logical operators are used to combine multiple conditions.

Python provides three logical operators:

- `and` → True when both conditions are True
- `or` → True when at least one condition is True
- `not` → reverses the Boolean result

````python
cgpa = 8.26
dsa_questions = 40

cgpa_eligible = cgpa >= 6.32
dsa_practice = dsa_questions >= 30

print("Both conditions satisfied:", cgpa_eligible and dsa_practice)
print("At least one condition satisfied:", cgpa_eligible or dsa_practice)
print("CGPA eligibility reversed:", not cgpa_eligible)
````

### Assignment Operators

Assignment operators are used to assign and update values stored in variables.

The basic assignment operator is `=`.
Python also provides shorthand operators such as `+=`, `-=`, `*=`, and `/=`.

````python
score = 50

print("Initial score:", score)

score += 10
print("After += 10:", score)

score -= 5
print("After -= 5:", score)

score *= 2
print("After *= 2:", score)

score /= 5
print("After /= 5:", score)
````

## 3. Conditional Statements

Conditional statements allow a program to make decisions based on whether a condition is `True` or `False`.

Python provides:

- `if`
- `elif`
- `else`

### 3.1 if Statement

The `if` statement executes a block of code when its condition evaluates to `True`.

````python
cgpa = 8.26

if cgpa >= 6.32:
    print("Student meets the CGPA requirement.")
````

### 3.2 if-else Statement

The `if-else` statement provides two possible execution paths.

- If the condition is `True`, the `if` block executes.
- If the condition is `False`, the `else` block executes.

````python
cgpa = 5.8

if cgpa >= 6.32:
    print("Student meets the CGPA requirement.")
else:
    print("Student does not meet the CGPA requirement.")
````

### 3.3 if-elif-else Statement

The `if-elif-else` structure is used when multiple conditions need to be evaluated.

- `if` checks the first condition.
- `elif` checks additional conditions if the previous conditions are False.
- `else` executes when none of the conditions are True.

````python
cgpa = 8.26

if cgpa >= 9.0:
    grade = "Excellent"
elif cgpa >= 8.0:
    grade = "Very Good"
elif cgpa >= 7.0:
    grade = "Good"
else:
    grade = "Needs Improvement"

print("CGPA:", cgpa)
print("Performance:", grade)
````

### 3.4 Nested if-else

A nested if-else statement is an `if-else` statement placed inside another `if` or `else` block.

It is useful when a decision depends on more than one level of conditions.

````python
cgpa = 8.26
has_required_skills = True

if cgpa >= 6.32:
    if has_required_skills:
        print("Student meets the basic internship criteria.")
    else:
        print("CGPA requirement is met, but required skills are missing.")
else:
    print("Student does not meet the CGPA requirement.")
````

### Key Takeaways

- `if` is used to execute code when a condition is true.
- `if-else` provides two possible execution paths.
- `if-elif-else` is useful when multiple conditions need to be evaluated.
- Nested `if-else` allows decisions to be made at multiple levels.
- Conditional statements form the foundation of decision-making in Python programs.

## 4. Loops

Loops are used to execute a block of code repeatedly.

They are useful when the same operation needs to be performed on multiple values, such as processing records in a dataset.

Python provides two main types of loops:

- `for` loop
- `while` loop

### 4.1 for Loop

A `for` loop is used to iterate over the elements of a sequence such as a list, tuple, string, or range.

````python
marks = [72, 85, 91, 68, 79]

for mark in marks:
    print("Mark:", mark)
````

### Processing Data with a for Loop

A `for` loop can be used to perform an operation on each value in a collection.

````python
marks = [72, 85, 91, 68, 79]

total = 0

for mark in marks:
    total += mark

print("Total marks:", total)
print("Number of students:", len(marks))
print("Average marks:", total / len(marks))
````

### Using range()

The `range()` function generates a sequence of numbers and is commonly used with `for` loops.

The starting value is included, while the ending value is excluded.

````python
for number in range(1, 6):
    print(number)
````

### break Statement

The `break` statement immediately terminates a loop when a specified condition is met.

````python
marks = [72, 85, 91, 68, 79]

for mark in marks:
    if mark >= 90:
        print("High score found:", mark)
        break
    print("Checking mark:", mark)
````

### continue Statement

The `continue` statement skips the current iteration and moves to the next iteration of the loop.

````python
marks = [72, 85, 91, 68, 79]

for mark in marks:
    if mark < 75:
        continue
    print("Eligible mark:", mark)
````

### Nested for Loop

A nested loop is a loop placed inside another loop.

The inner loop executes completely for each iteration of the outer loop.

````python
students = ["Student A", "Student B", "Student C"]
subjects = ["Python", "Statistics"]

for student in students:
    for subject in subjects:
        print(student, "-", subject)
````

### 4.7 while Loop

A `while` loop repeatedly executes a block of code as long as a specified condition remains `True`.

Unlike a `for` loop, a `while` loop is generally used when the number of iterations depends on a condition.


````python
count = 1

while count <= 5:
    print("Iteration:", count)
    count += 1
````

### Key Takeaways

- `for` loops are useful for iterating over sequences and collections.
- `while` loops continue execution while a condition remains `True`.
- `range()` generates a sequence of numbers commonly used with `for` loops.
- `break` terminates a loop immediately.
- `continue` skips the current iteration and moves to the next one.
- Nested loops are useful when working with multiple levels of data.
- Loops provide the foundation for repetitive data-processing tasks.

## 5. Functions

A function is a reusable block of code designed to perform a specific task.

Functions help to:

- Reduce code repetition
- Improve code organization
- Make programs easier to understand and maintain
- Reuse logic with different inputs

### 5.1 Defining a Function

A function is defined using the `def` keyword followed by the function name and parentheses.

Basic syntax:

```python
def function_name():
    # statements

````python

def greet():
    print("Welcome to Data Science!")

greet()
````

### 5.2 Parameters and Arguments

A parameter is a variable defined in a function.

An argument is the actual value passed to the function when it is called.

Parameters allow a function to work with different input values.

````python
def calculate_square(number):
    print("Square:", number ** 2)

calculate_square(5)
calculate_square(10)
````

### 5.3 Return Values

The `return` statement sends a result back from a function to the place where the function was called.

Returning a value is useful when the result needs to be stored, reused, or passed to another operation.

````python
def calculate_average(marks):
    total = sum(marks)
    average = total / len(marks)
    return average

student_marks = [72, 85, 91, 68, 79]

average_marks = calculate_average(student_marks)

print("Average marks:", average_marks)
````

### 5.4 Multiple Parameters

A function can accept multiple parameters to perform a task using more than one input.

This makes functions flexible and reusable for different situations.


````python
def calculate_percentage(obtained_marks, total_marks):
    percentage = (obtained_marks / total_marks) * 100
    return percentage

percentage = calculate_percentage(425, 500)

print("Percentage:", percentage)
````

### 5.5 Default Arguments

A default argument is a parameter that already has a predefined value.

If the caller does not provide a value for that parameter, Python uses the default value.

````python
def calculate_percentage(obtained_marks, total_marks=500):
    percentage = (obtained_marks / total_marks) * 100
    return percentage

print("Percentage:", calculate_percentage(425))
print("Percentage:", calculate_percentage(850, 1000))
````

### 5.6 Function Scope

Scope determines where a variable can be accessed in a Python program.

A variable created inside a function normally has **local scope**, while a variable created outside a function has **global scope**.

````python
course = "Data Science"

def show_course():
    topic = "Python Functions"
    print("Course:", course)
    print("Current Topic:", topic)

show_course()
````

````python
def create_variable():
    message = "This is a local variable"
    print(message)

create_variable()
````

### 5.7 Practical Example: Student Performance Analysis

A reusable function can combine multiple operations and return meaningful information from data.

The following function calculates the total marks, average marks, and performance category for a student.

````python
def analyze_performance(marks):
    total = sum(marks)
    average = total / len(marks)

    if average >= 85:
        performance = "Excellent"
    elif average >= 70:
        performance = "Good"
    elif average >= 50:
        performance = "Average"
    else:
        performance = "Needs Improvement"

    return total, average, performance


student_marks = [78, 85, 92, 74, 81]

total, average, performance = analyze_performance(student_marks)

print("Total Marks:", total)
print("Average Marks:", average)
print("Performance:", performance)
````

### Key Takeaways

- Functions are reusable blocks of code designed for specific tasks.
- Parameters allow functions to accept input values.
- Arguments are the actual values passed to parameters.
- The `return` statement sends a result back to the caller.
- Multiple parameters can be used when a function requires multiple inputs.
- Default arguments provide predefined values when an argument is not supplied.
- Variables inside functions generally have local scope.
- Functions help break complex problems into smaller, reusable components.
- Combining functions with conditions and data structures enables practical data-processing workflows.
